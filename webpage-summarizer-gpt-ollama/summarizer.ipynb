### Add system and User Prompts

```python
system_prompt = """
you are a helpful assistant that can analyze the contents of a website, and provides a short summary, ignoring text that might be navigation related.
respond in markdown format. Do not wrap the markdown in a code block - respond just with the markdown.
"""

user_prompt_prefix = """
Here are the contents of a website.
Provide a short summary of this website.
If it includes news or announcements, then summarize these too.
"""
```

### functions to call Scraper.py with GPT 

```python
openai = OpenAI()

def messages_for(website):
    return [
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": user_prompt_prefix + website}
    ]

def summarize(url):
    website = fetch_website_contents(url)
    response = openai.chat.completions.create(
        model = "gpt-4.1-mini",
        messages= messages_for(website)
    )
    return response.choices[0].message.content

def display_summary(url):
    summary = summarize(url)
    display(Markdown(summary))
```

```python
display_summary("https://www.anthropic.com/")
```

#### Run the summerize function with ollama 

```python
OLLAMA_BASE_URL = "http://localhost:11434/v1"

ollama = OpenAI(base_url = OLLAMA_BASE_URL, api_key= "ollama")

def ollama_summarize(url):
    website = fetch_website_contents(url)
    response = ollama.chat.completions.create(
        model = "llama3.2",
        messages = messages_for(website)
    )
    return response.choices[0].message.content

def display_summary_ollama(url):
    summary = ollama_summarize(url)
    display(Markdown(summary))
```

```python
display_summary_ollama("https://www.anthropic.com/")
```
